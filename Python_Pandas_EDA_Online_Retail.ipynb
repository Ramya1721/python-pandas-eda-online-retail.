```python
import pandas as pd
import matplotlib.pyplot as plt
d= pd.read_csv('/content/online_retail.csv', encoding='ISO-8859-1')
print("Dataset loaded successfully!")
print("Shape:", d.shape)
display(d.head())
```


# Python & Pandas Exploratory Data Analysis (EDA)
## Dataset: Online Retail
This project performs exploratory data analysis on an online retail transaction dataset using Python, Pandas, Matplotlib, and Seaborn.
The objective is to understand the dataset structure, identify missing values, analyse summary statistics, explore patterns through visualisations, and derive meaningful insights.

```python
# Dataset dimensions
print("Number of rows:",d.shape[0])
print("Number of columns:",d.shape[1])
# Column names
print("\nColumn names:")
print(d.columns.tolist())
print("\nDataset information:")
d.info()
print("\nMemory usage:")
print(f"{d.memory_usage(deep=True).sum() / (1024**2):.2f} MB")
print("\nFirst five rows:")
display(d.head())
print("\nLast five rows:")
display(d.tail())
```

```python
# Count missing values in each column
missing = d.isnull().sum()
missing_percent = (missing / len(d)) * 100
missing_table = pd.DataFrame({
    'Missing Values': missing,
    'Missing Percentage': missing_percent
})
print("Missing Value Analysis:")
display(missing_table)
```

```python
# Summary statistics for numerical columns
print("Numerical Summary Statistics:")
display(d.describe())
print("\nMedian Values:")
display(d.median(numeric_only=True))
# Most frequent countries
print("Top 10 Countries by Transaction Rows:")
display(d['Country'].value_counts().head(10))
# Most frequent products
print("\nTop 10 Product Descriptions:")
display(d['Description'].value_counts().head(10))
# Most frequent invoice numbers
print("\nNumber of Unique Invoices:")
print(d['InvoiceNo'].nunique())
# Number of unique products
print("\nNumber of Unique Products:")
print(d['StockCode'].nunique())
```

**Visualisation 1: Distribution of Transaction Quantities**

**Question:** What quantities do customers typically purchase in a transaction row?

```python
plt.figure(figsize=(9, 5))
d['Quantity'].clip(-100, 100).hist(bins=50)
plt.title('Distribution of Transaction Quantities')
plt.xlabel('Quantity (clipped to -100 and 100)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

**Interpretation**:The histogram shows how transaction quantities are distributed. The values are clipped for visualisation so that extreme quantities do not dominate the chart; negative quantities may represent returns or cancellations.

**Visualisation 2:** Top 10 Countries by Transaction Count

**Question:** Which countries have the most transaction records?

```python
country_counts = d['Country'].value_counts().head(10)
plt.figure(figsize=(10, 5))
country_counts.plot(kind='bar')
plt.title('Top 10 Countries by Transaction Count')
plt.xlabel('Country')
plt.ylabel('Number of Transaction Rows')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

**Interpretation**: This chart compares the number of transaction rows recorded for the top 10 countries. The country with the tallest bar contributes the most rows, but this does not necessarily mean it has the highest revenue.

**Visualisation 3**: Distribution of Unit Prices

**Question**: How are product prices distributed?

```python
prices = d.loc[d['UnitPrice'] > 0, 'UnitPrice']
plt.figure(figsize=(9, 5))
prices.clip(upper=prices.quantile(0.99)).hist(bins=50)
plt.title('Distribution of Unit Prices')
plt.xlabel('Unit Price (upper-clipped at 99th percentile)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

**Interpretation:** This histogram displays the distribution of positive unit prices, with extreme values clipped at the 99th percentile for readability.A concentration of observations at lower prices would suggest that many transaction rows involve relatively inexpensive items.

**Visualisation 4:** Monthly Transaction Trends

**Question:** How does transaction activity change over time?

```python
d['InvoiceDate'] = pd.to_datetime(d['InvoiceDate'])
monthly_transactions = (
    d.set_index('InvoiceDate')
     .resample('MS')
     .size()
)
plt.figure(figsize=(11, 5))
monthly_transactions.plot(kind='line', marker='o')
plt.title('Monthly Transaction Row Count')
plt.xlabel('Month')
plt.ylabel('Number of Transaction Rows')
plt.tight_layout()
plt.show()
```

**Interpretation:** The line chart shows how transaction-row volume changes month by month. Peaks and dips indicate changes in recorded activity, though they do not necessarily reflect changes in revenue or unique customers.

**Visualisation 5:** Correlation Between Numerical Variables

**Question:** Are quantity and unit price related?

```python
numeric_data = d[['Quantity', 'UnitPrice']]
correlation = numeric_data.corr()
plt.figure(figsize=(6, 4))
plt.imshow(correlation, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar(label='Correlation Coefficient')
plt.xticks(range(2), correlation.columns)
plt.yticks(range(2), correlation.columns)
plt.title('Correlation Between Quantity and Unit Price')
for i in range(2):
    for j in range(2):
        plt.text(j, i, f'{correlation.iloc[i, j]:.2f}',
                 ha='center', va='center')
plt.tight_layout()
plt.show()
```

**Interpretation:** The correlation matrix measures the linear relationship between quantity and unit price. A value close to zero indicates a weak linear relationship, while a positive or negative value indicates the direction of the relationship; correlation does not establish causation.

**Visualisation 6:** Top 10 Most Frequently Listed Products

**Question**: Which product descriptions appear most often?

```python
top_products = d['Description'].value_counts().head(10).sort_values()
plt.figure(figsize=(10, 6))
top_products.plot(kind='barh')
plt.title('Top 10 Most Frequently Listed Products')
plt.xlabel('Number of Transaction Rows')
plt.ylabel('Product Description')
plt.tight_layout()
plt.show()
```

**Interpretation:** This chart identifies the product descriptions that occur most frequently in transaction records. Frequent appearances do not necessarily mean these products generated the highest revenue or had the greatest quantity sold.

**Key Findings.**

```python
print("Total transaction rows:", len(d))
print("Unique invoices:", d['InvoiceNo'].nunique())
print("Unique products:", d['StockCode'].nunique())
print("Countries represented:", d['Country'].nunique())
print("\nCountry with the most transaction rows:")
print(d['Country'].value_counts().idxmax())
print("\nAverage quantity:", round(d['Quantity'].mean(), 2))
print("Median quantity:", d['Quantity'].median())
print("\nAverage unit price:", round(d['UnitPrice'].mean(), 2))
print("Median unit price:", d['UnitPrice'].median())
print("\nMissing CustomerID percentage:", round(d['CustomerID'].isnull().mean() * 100, 2), "%")
```
